```python
import os
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from IPython.display import display_markdown
from langchain_core.messages import HumanMessage , SystemMessage , AIMessage 

load_dotenv()

llm = ChatGroq(model = "llama-3.1-8b-instant",api_key = os.getenv("GROQ_API_KEY"))
```

```python
generation_chat_history = [
    SystemMessage(content="""You are a Python Programmer tasked with generating high quality Python Code.
                  If the user provides critique , respond with a revised version of your previous attempt""")
]
generation_chat_history.append(HumanMessage(content="Generate a Python Implementation of the merge sort algorithm"))
mergesort_code = llm.invoke(generation_chat_history)
display_markdown(mergesort_code.content , raw = True)
```

```python
reflection_chat_history = [
    SystemMessage(content="You are Andrej Karpathy , an experienced computer scientist. You are tasked with generating critique and recommendation of python code , dont give me code just give me recommendation and critques")
]
reflection_chat_history.append(HumanMessage(content="mergesort_code"))
critique = llm.invoke(reflection_chat_history)
display_markdown(critique.content,raw=True)
```

```python
generation_chat_history.append(HumanMessage(content=critique.content))
essay = llm.invoke(generation_chat_history)
display_markdown(essay.content,raw=True)
```
